## Enrich data

**Enrich Data**

*In this step, we enhance the existing datasets by enriching them with derived features and user-specific plan details. This enrichment will help analyze the total revenue and usage patterns for each user, which will be critical in comparing the two plans (Surf and Ultimate).*

---

**Actions Performed:**

1. **Mapped Plan Details to Users:**

   * Mapped plan-specific details like `mb_per_month_included`, `minutes_included`, and `usd_per_minute` from the `plans_data` DataFrame to the `users_data` DataFrame based on each user's plan.
   * This step allows us to calculate costs more effectively for each user.


2. **Enriched Messages Data:**

   * Ensured the presence of a `messages` column, even if it was missing, by counting the records for each `user_id`.
   * Merged `messages_data` with the `users_data` to include the `usd_per_message` column for cost calculations.
   * Calculated `messages_cost` by multiplying the total messages sent by the cost per message.


3. **Enriched Internet Data:**

   * Merged `internet_data` with `users_data` to include plan-specific internet limits and costs (e.g., `mb_per_month_included`, `usd_per_gb`).
   * Calculated extra internet usage (`extra_gb`) exceeding the monthly limit.
   * Calculated the cost for extra internet usage (`internet_cost`).


4. **Enriched Calls Data:**

   * Merged `calls_data` with `users_data` to include plan-specific call limits and costs.
   * Calculated extra minutes (`extra_minutes`) exceeding the monthly limit.
   * Calculated the cost for extra call usage (`calls_cost`).


5. **Calculated Total Revenue:**

   * Aggregated revenue from messages, calls, internet usage, and monthly fees for each user.


6. **Validation and Display:**

   * Displayed the enriched `users_data` DataFrame to verify the additions and correctness of calculations.

---

**Insights:**

* Enrichment provides a comprehensive view of the costs associated with each user's activity.
* This detailed breakdown enables analysis of revenue trends and profitability for each plan.




```python
plans_mapping = plans_data.set_index('plan_name').to_dict('index')

users_data['mb_per_month_included'] = users_data['plan'].map(lambda x: plans_mapping[x]['mb_per_month_included'])
users_data['minutes_included'] = users_data['plan'].map(lambda x: plans_mapping[x]['minutes_included'])
users_data['usd_monthly_pay'] = users_data['plan'].map(lambda x: plans_mapping[x]['usd_monthly_pay'])
users_data['usd_per_gb'] = users_data['plan'].map(lambda x: plans_mapping[x]['usd_per_gb'])
users_data['usd_per_message'] = users_data['plan'].map(lambda x: plans_mapping[x]['usd_per_message'])
users_data['usd_per_minute'] = users_data['plan'].map(lambda x: plans_mapping[x]['usd_per_minute'])

if 'messages' not in messages_data.columns:
    
    messages_data['messages'] = messages_data.groupby('user_id')['id'].transform('count')

messages_data = messages_data.merge(
    users_data[['user_id', 'usd_per_message']],
    on='user_id',
    how='left'
)

messages_data['messages_cost'] = messages_data['messages'] * messages_data['usd_per_message']

internet_data = internet_data.merge(
    users_data[['user_id', 'mb_per_month_included', 'usd_per_gb']],
    on='user_id',
    how='left'
)

internet_data['extra_gb'] = (internet_data['mb_used'] - internet_data['mb_per_month_included']) / 1024
internet_data['extra_gb'] = internet_data['extra_gb'].clip(lower=0)
internet_data['internet_cost'] = internet_data['extra_gb'] * internet_data['usd_per_gb']

calls_data = calls_data.merge(
    users_data[['user_id', 'minutes_included', 'usd_per_minute']],
    on='user_id',
    how='left'
)

calls_data['extra_minutes'] = calls_data['duration'] - calls_data['minutes_included']
calls_data['extra_minutes'] = calls_data['extra_minutes'].clip(lower=0)
calls_data['calls_cost'] = calls_data['extra_minutes'] * calls_data['usd_per_minute']

total_revenue = (
    messages_data.groupby('user_id')['messages_cost'].sum() +
    internet_data.groupby('user_id')['internet_cost'].sum() +
    calls_data.groupby('user_id')['calls_cost'].sum() +
    users_data.set_index('user_id')['usd_monthly_pay']
)

users_data['average_call_duration'] = calls_data.groupby('user_id')['duration'].mean()
users_data['total_messages'] = messages_data.groupby('user_id')['messages'].sum()
users_data['total_mb_used'] = internet_data.groupby('user_id')['mb_used'].sum()

users_data['total_revenue'] = total_revenue

print("Enriched Users Data:")
display(users_data.head())
```

### Enrich data

To enhance the calls dataset for deeper analysis, we will add the following derived features:

- **Call Costs**:
  - Calculate the total call costs for users exceeding their plan's included minutes. This requires merging with the users' dataset to bring in plan-specific limits and rates.

These additional features will help in analyzing user behavior and evaluating plan profitability.


```python
import numpy as np

plans_mapping = {
    'surf': {'minutes_included': 500, 'usd_per_minute': 0.03},
    'ultimate': {'minutes_included': 3000, 'usd_per_minute': 0.01}
}

calls_data = calls_data.merge(
    users_data[['user_id', 'plan', 'usd_per_minute']],
    on='user_id',
    how='left',
    suffixes=("", "_user")
)

if 'plan' in calls_data.columns:
    calls_data['extra_minutes'] = (
        np.ceil(calls_data['duration']) - 
        calls_data['plan'].map(lambda x: plans_mapping[x]['minutes_included']).clip(lower=0)
    )
    
    calls_data['call_cost'] = calls_data['extra_minutes'].clip(lower=0) * calls_data['usd_per_minute']

    call_summary = calls_data.groupby('user_id').agg({
        'id': 'count',
        'duration': 'sum',
        'call_cost': 'sum'
    }).reset_index().rename(columns={
        'id': 'total_calls',
        'duration': 'total_call_duration',
        'call_cost': 'total_call_cost'
    })

    print(call_summary.head())
else:
    print("Error: 'plan' column is missing after merge.")
```

## Aggregate data per user

In this section, we aggregate the cleaned data to summarize usage metrics for each user per month. This includes the number of calls, total call duration, number of messages, and internet traffic volume. Once aggregated, the data is combined into a single DataFrame for further analysis, including plan-based calculations for monthly revenue.

```python
# Calculate the number of calls made by each user per month. Save the result.
calls_per_user = calls_data.groupby(['user_id', calls_data['call_date'].dt.month])['id'].count().reset_index()
calls_per_user.columns = ['user_id', 'month', 'total_calls']
print("Number of Calls per User per Month:")
display(calls_per_user.head())
```

```python
# Calculate the amount of minutes spent by each user per month. Save the result.
minutes_per_user = calls_data.groupby(['user_id', calls_data['call_date'].dt.month])['duration'].sum().reset_index()
minutes_per_user.columns = ['user_id', 'month', 'total_minutes']
print("Minutes Spent per User per Month:")
display(minutes_per_user.head())
```

```python
# Calculate the number of messages sent by each user per month. Save the result.
messages_per_user = messages_data.groupby(['user_id', messages_data['message_date'].dt.month])['id'].count().reset_index()
messages_per_user.columns = ['user_id', 'month', 'total_messages']
print("Number of Messages per User per Month:")
display(messages_per_user.head())
```

```python
# Calculate the volume of internet traffic used by each user per month. Save the result.
internet_per_user = internet_data.groupby(['user_id', internet_data['session_date'].dt.month])['mb_used'].sum().reset_index()
internet_per_user.columns = ['user_id', 'month', 'total_mb_used']
print("Internet Traffic per User per Month:")
display(internet_per_user.head())
```

To analyze user behavior effectively, it is necessary to combine the aggregated data for calls, minutes, messages, and internet usage into a single DataFrame. This allows us to have a consolidated view of each user’s total usage metrics for each month.

**Steps:**


1. Aggregate individual metrics (total calls, total minutes, total messages, and total internet data) grouped by user ID and month.
2. Merge the separate metrics into one DataFrame using the user ID and month as keys.
3. Replace missing values (if any) with `0`, as missing data indicates no usage for that category in the respective month.


This combined DataFrame will serve as the foundation for further calculations, including revenue computation and plan comparison.


```python
# Merge the data for calls, minutes, messages, internet based on user_id and month
user_monthly_data = calls_per_user.merge(minutes_per_user, on=['user_id', 'month'], how='outer')
user_monthly_data = user_monthly_data.merge(messages_per_user, on=['user_id', 'month'], how='outer')
user_monthly_data = user_monthly_data.merge(internet_per_user, on=['user_id', 'month'], how='outer')

user_monthly_data.fillna(0, inplace=True)
print("Aggregated User Data per Month:")
display(user_monthly_data.head())
```

```python
# Add the plan information
user_monthly_data = user_monthly_data.merge(users_data[['user_id', 'plan']], on='user_id', how='left')
print("User Data with Plan Information:")
display(user_monthly_data.head())
```

To calculate monthly revenue:

1. **Identify Overage Usage:**
   - Subtract the free package limits for calls, messages, and internet usage from the total usage in each category.
   - For overage calculations:
     - Calls exceeding the included limit are charged per minute.
     - Messages exceeding the included limit are charged per message.
     - Internet data exceeding the included limit is charged per GB.


2. **Compute Overage Costs:**
   - Multiply the overage values for each category by the corresponding per-unit charges.


3. **Calculate Total Revenue:**
   - Add the monthly fixed subscription fee (based on the user's plan) to the computed overage costs.


4. **Create a New Revenue Column:**
   - Store the total monthly revenue for each user in the consolidated DataFrame for further analysis.


```python
# Calculate the monthly revenue for each user

plans_mapping = plans_data.set_index('plan_name').to_dict('index')

def calculate_revenue(row):
    plan = plans_mapping[row['plan']]
    # Calculate overage for calls, messages, and internet
    extra_minutes = max(row['total_minutes'] - plan['minutes_included'], 0)
    extra_messages = max(row['total_messages'] - plan['messages_included'], 0)
    extra_gb = max(row['total_mb_used'] - plan['mb_per_month_included'], 0) / 1024
   
    extra_cost = (extra_minutes * plan['usd_per_minute'] +
                  extra_messages * plan['usd_per_message'] +
                  extra_gb * plan['usd_per_gb'])
    
    return plan['usd_monthly_pay'] + extra_cost

user_monthly_data['monthly_revenue'] = user_monthly_data.apply(calculate_revenue, axis=1)

print("Monthly Revenue per User:")
display(user_monthly_data[['user_id', 'month', 'monthly_revenue']].head())
```

```python
calls_agg = calls_data.groupby(["user_id", "month"]).agg({"duration": "sum"}).reset_index()
calls_agg.rename(columns={"duration": "total_minutes"}, inplace=True)

messages_agg = messages_data.groupby(["user_id", "month"]).agg({"messages": "sum"}).reset_index()
messages_agg.rename(columns={"messages": "total_messages"}, inplace=True)

internet_agg = internet_data.groupby(["user_id", "month"]).agg({"mb_used": "sum"}).reset_index()
internet_agg.rename(columns={"mb_used": "total_mb"}, inplace=True)

usage_data = calls_agg.merge(messages_agg, on=["user_id", "month"], how="outer")
usage_data = usage_data.merge(internet_agg, on=["user_id", "month"], how="outer")
usage_data.fillna(0, inplace=True)

revenue_data = users_data.merge(usage_data, on="user_id", how="left")
revenue_data["month"] = revenue_data["reg_date"].dt.month

revenue_data.head()
```

```python
plans_data.rename(columns={"usd_monthly_pay": "usd_per_month"}, inplace=True)

def calculate_revenue(row):
    fixed_cost = plans_data.loc[plans_data['plan_name'] == row['plan'], 'usd_per_month'].values[0]
    extra_minutes_cost = max(0, row['total_minutes'] - plans_data.loc[plans_data['plan_name'] == row['plan'], 'minutes_included'].values[0]) * plans_data.loc[plans_data['plan_name'] == row['plan'], 'usd_per_minute'].values[0]
    extra_messages_cost = max(0, row['total_messages_y'] - plans_data.loc[plans_data['plan_name'] == row['plan'], 'messages_included'].values[0]) * plans_data.loc[plans_data['plan_name'] == row['plan'], 'usd_per_message'].values[0]
    extra_mb_cost = max(0, row['total_mb'] - plans_data.loc[plans_data['plan_name'] == row['plan'], 'mb_per_month_included'].values[0]) * plans_data.loc[plans_data['plan_name'] == row['plan'], 'usd_per_gb'].values[0] / 1024
    return fixed_cost + extra_minutes_cost + extra_messages_cost + extra_mb_cost

revenue_data['monthly_revenue'] = revenue_data.apply(calculate_revenue, axis=1)

simplified_output = revenue_data[['user_id', 'plan', 'month', 'total_minutes', 'total_messages_y', 'total_mb', 'monthly_revenue']]
print(simplified_output.head())
```

## Test statistical hypotheses

1. **Null Hypothesis (H₀):** The average monthly revenue for Surf and Ultimate plans is the same.  
2. **Alternative Hypothesis (H₁):** The average monthly revenue for Surf and Ultimate plans differs.  
3. **Significance Level (α):** 0.05  
4. **Test Used:** Two-sample t-test for independent samples.  

<div class="alert alert-block alert-success">✔️
    

__Reviewer's comment №1__

Correct interpretation of the null and alternative hypothesis

- **Result of the Test:**  
  - T-test statistic: **9.10**  
  - P-value: **0.0000**  

- **Conclusion:**  
  Since the p-value is less than the significance level (0.05), we **reject the null hypothesis**.  
  The average revenue for users of the Surf and Ultimate plans **differs significantly**.


```python
from scipy.stats import ttest_ind

surf_revenue = revenue_data[revenue_data['plan'] == 'surf']['monthly_revenue']
ultimate_revenue = revenue_data[revenue_data['plan'] == 'ultimate']['monthly_revenue']

alpha = 0.05

stat, p_value = ttest_ind(surf_revenue, ultimate_revenue, equal_var=False)

print(f"T-test statistic: {stat:.2f}")
print(f"P-value: {p_value:.4f}")

if p_value < alpha:
    print("We reject the null hypothesis: The average revenue for Surf and Ultimate plans differs.")
else:
    print("We fail to reject the null hypothesis: No significant difference in average revenue between the plans.")
```
