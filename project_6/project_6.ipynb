# **14. Hypothesis Testing: Rainy Saturdays Impact on Ride Duration**  

We perform a t-test to determine if ride duration on rainy Saturdays is significantly different from other days.

```python
alpha = 0.05
t_stat, p_value = st.ttest_ind(bad_weather, good_weather, equal_var=False)

print(f"P-value: {p_value}")

if p_value < alpha:
    print("We reject the null hypothesis: Rainy Saturdays significantly impact ride durations.")
else:
    print("We fail to reject the null hypothesis: No significant impact of rain on ride durations.")
```

# **Objective:**
This hypothesis test determines whether **rainy Saturdays significantly impact ride durations** compared to other days with good weather.

# **Hypothesis Formulation:**
- **Null Hypothesis (H₀):** There is **no significant difference** in ride duration between **rainy Saturdays** and other days.
- **Alternative Hypothesis (H₁):** Ride durations on **rainy Saturdays are significantly different** from those on other days.

# **Code Breakdown:**
- **Significance Level (`alpha`) is set to 0.05.**
- A **t-test for independent samples** (`st.ttest_ind`) is performed:
  - `bad_weather` represents ride durations on rainy Saturdays.
  - `good_weather` represents ride durations on other days.
  - `equal_var=False` assumes that the two groups **do not necessarily have equal variance**.
- The **p-value is compared to alpha** to determine if we reject or fail to reject the null hypothesis.

# **Findings:**
- **P-value = 2.86e-07**, which is **significantly smaller than alpha (0.05)**.
- Since **p-value < 0.05**, we **reject the null hypothesis**.
- **Conclusion:** **Rainy Saturdays significantly impact ride durations.**
  - This suggests that rain likely causes **longer travel times**, possibly due to **traffic congestion, slower driving speeds, or adverse road conditions**.

- **Compare ride durations by time of day** to see if morning or evening trips are more affected by rain.
- **Investigate other weather conditions** (e.g., snow, fog) to understand their impact on ride times.
